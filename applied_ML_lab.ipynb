```python
#implementation of the quick sort algorithm in python [3,6,8,10,1,2,1]

def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    else:
        pivot = arr[len(arr) // 2]
        left = [x for x in arr if x < pivot]
        middle = [x for x in arr if x == pivot]
        right = [x for x in arr if x > pivot]
        return quick_sort(left) + middle + quick_sort(right)

my_list_to_sort = [3, 6, 8, 10, 1, 2, 1]
sorted_list = quick_sort(my_list_to_sort)
print(f"Original : {my_list_to_sort}")
print(f"Sorted : {sorted_list}")
```

```python
import numpy as np
#9.
#Do not use a Python for loop.
#Given:
marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])
print("Original marks array:\n", marks)

#Using NumPy boolean indexing, find:
#Marks greater than 75
marks_gt_75 = marks[marks > 75]
print("\nMarks greater than 75:", marks_gt_75)

#Marks between 50 and 80
marks_between_50_and_80 = marks[(marks >= 50) & (marks <= 80)]
print("\nMarks between 50 and 80:", marks_between_50_and_80)

#Number of students who scored below 50
students_below_50 = marks[marks < 50]
num_students_below_50 = len(students_below_50)
print("\nNumber of students who scored below 50:", num_students_below_50)

#Percentage of students who scored 60 or above
students_60_or_above = marks[marks >= 60]
percentage_60_or_above = (len(students_60_or_above) / len(marks)) * 100
print("\nPercentage of students who scored 60 or above: {:.2f}%".format(percentage_60_or_above))
```

```python
import numpy as np

#10. The following array contains marks of four students in three subjects:
marks = np.array([
   [80, 75, 90],
   [65, 70, 72],
   [88, 92, 85],
   [55, 60, 58]
])

print("Original marks array:\n", marks)

# Calculate:
# Total marks of each student
total_per_student = np.sum(marks, axis=1)
print("\nTotal marks:", total_per_student)

# Average marks of each student
average_per_student = np.mean(marks, axis=1)
print("\nAverage marks:", average_per_student)

# Average marks in each subject
average_per_subject = np.mean(marks, axis=0)
print("\nAverage marks in each subject:", average_per_subject)

# Student with the highest total
highest_total_student_index = np.argmax(total_per_student)
highest_total_student_marks = marks[highest_total_student_index]
print("\nStudent with the highest total marks (index {}): {}".format(highest_total_student_index, highest_total_student_marks))
print("Highest total marks:", total_per_student[highest_total_student_index])

# Highest mark in each subject.
highest_mark_in_each_subject = np.max(marks, axis=0)
print("\nHighest mark in each subject:", highest_mark_in_each_subject)
```

```python
import numpy as np

#11. Create:
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:\n", A)
print("Matrix B:\n", B)

#Perform:
# Matrix addition
matrix_addition = A + B
print("\nMatrix Addition (A + B):\n", matrix_addition)

# Element-wise multiplication
elementwise_multiplication = A * B
print("\nElement-wise Multiplication (A * B):\n", elementwise_multiplication)

# Matrix multiplication
matrix_multiplication = A @ B
print("\nMatrix Multiplication (A @ B):\n", matrix_multiplication)

# Transpose of A
transpose_A = A.T
print("\nTranspose of A:\n", transpose_A)

# Explain the difference between: A * B and A @ B
print("\nExplanation of A * B vs A @ B:")
print("A * B performs element-wise multiplication, meaning each element of A is multiplied by the corresponding element of B.")
print("A @ B performs matrix multiplication (dot product). This operation follows the rules of linear algebra")
```

```python
import numpy as np

# 12. Daily temperatures for two weeks are:

temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])

print("Daily temperatures:\n", temperature)

# Find:]
# Mean temperature
mean_temp = np.mean(temperature)
print("\nMean temperature: {:.2f}°C".format(mean_temp))

# Maximum and minimum temperatures
max_temp = np.max(temperature)
min_temp = np.min(temperature)
print("Maximum temperature: {}°C".format(max_temp))
print("Minimum temperature: {}°C".format(min_temp))

# Number of days above the average temperature
days_above_average = np.sum(temperature > mean_temp)
print("Number of days above the average temperature: {}".format(days_above_average))

# Temperatures greater than 34°C
temps_gt_34 = temperature[temperature > 34]
print("Temperatures greater than 34°C: {}".format(temps_gt_34))

# Difference between maximum and minimum
diff_max_min = max_temp - min_temp
print("Difference between maximum and minimum temperature: {}°C".format(diff_max_min))

# Standard deviation
std_dev_temp = np.std(temperature)
print("Standard deviation of temperature: {:.2f}°C".format(std_dev_temp))
```

```python
#13. Monthly sales of a company are:
#sales = np.array([
#    12000, 15000, 17000, 14000,
#    18000, 21000, 19000, 22000,
#    25000, 23000, 26000, 28000
#])

#Determine:
#Total annual sales
#Average monthly sales
#Month with the highest sales
#Month with the lowest sales
#Number of months with sales above ₹20,000
#Percentage increase from January to December
#Hint: Remember that NumPy indexing starts at 0, while month numbering normally starts at 1.


import numpy as np

sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])

print("Monthly Sales:\n", sales)

# Total annual sales
total_annual_sales = np.sum(sales)
print("\nTotal annual sales: ₹{:,.2f}".format(total_annual_sales))

# Average monthly sales
average_monthly_sales = np.mean(sales)
print("Average monthly sales: ₹{:,.2f}".format(average_monthly_sales))

# Month with the highest sales
highest_sales_month_index = np.argmax(sales)
highest_sales_value = sales[highest_sales_month_index]
# Add 1 to index to get the actual month number (1-indexed)
print("Month with the highest sales: Month {} (₹{:,.2f})".format(highest_sales_month_index + 1, highest_sales_value))

# Month with the lowest sales
lowest_sales_month_index = np.argmin(sales)
lowest_sales_value = sales[lowest_sales_month_index]
# Add 1 to index to get the actual month number (1-indexed)
print("Month with the lowest sales: Month {} (₹{:,.2f})".format(lowest_sales_month_index + 1, lowest_sales_value))

# Number of months with sales above ₹20,000
months_above_20k = np.sum(sales > 20000)
print("Number of months with sales above ₹20,000: {}".format(months_above_20k))

# Percentage increase from January to December
january_sales = sales[0]
december_sales = sales[-1]

if january_sales != 0:
    percentage_increase = ((december_sales - january_sales) / january_sales) * 100
    print("Percentage increase from January to December: {:.2f}%".format(percentage_increase))
else:
    print("Cannot calculate percentage increase from January to December (January sales are zero).")
```

```python
#21. Take your own example to normalize an array.
# Min-Max Normalization: (x - min(x)) / (max(x) - min(x))

my_array_to_normalize = np.array([10, 20, 30, 40, 50, 100])
print("Original array for normalization:", my_array_to_normalize)

min_val = np.min(my_array_to_normalize)
max_val = np.max(my_array_to_normalize)

normalized_array = (my_array_to_normalize - min_val) / (max_val - min_val)
print("Normalized array:", normalized_array)
```

```python
#22. Use your own data to standardize an array.
# Standardization: (x - mean(x)) / std(x)

my_array_to_standardize = np.array([5, 10, 15, 20, 25, 30])
print("Original array for standardization:", my_array_to_standardize)

mean_val = np.mean(my_array_to_standardize)
std_dev = np.std(my_array_to_standardize)

standardized_array = (my_array_to_standardize - mean_val) / std_dev
print("Standardized array:", standardized_array)
```

```python
#24. Compare the time required to multiply 1,000,000 numbers using a Python list and a NumPy array.
import time

SIZE = 1_000_000

# Python list multiplication
python_list1 = list(range(SIZE))
python_list2 = list(range(SIZE))

start_time = time.time()
python_result = [a * b for a, b in zip(python_list1, python_list2)]
end_time = time.time()
python_time = end_time - start_time
print(f"Time taken for Python list multiplication: {python_time:.6f} seconds")

# NumPy array multiplication
numpy_array1 = np.arange(SIZE)
numpy_array2 = np.arange(SIZE)

start_time = time.time()
numpy_result = numpy_array1 * numpy_array2
end_time = time.time()
numpy_time = end_time - start_time
print(f"Time taken for NumPy array multiplication: {numpy_time:.6f} seconds")

print(f"\nNumPy is approximately {python_time / numpy_time:.2f} times faster than Python lists for this operation.")
```

### 15. Find duplicates, display duplicate rows, and remove duplicates.

```python
import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Meena", "Asha", "Arun"],
    "Marks": [85, 78, 92, 85, 88]
}

df2 = pd.DataFrame(data)
print("Original DataFrame:\n", df2)

# Find duplicates
duplicates = df2.duplicated()
print("\nBoolean Series indicating duplicate rows:\n", duplicates)

# Display duplicate rows
duplicate_rows = df2[df2.duplicated()]
print("\nDuplicate rows (based on all columns):\n", duplicate_rows)

# Remove duplicates
df_no_duplicates = df2.drop_duplicates()
print("\nDataFrame after removing duplicates:\n", df_no_duplicates)
```

### 16. Calculate: Mean, Median, Maximum, Minimum, Standard deviation, Sum for the given DataFrame.

```python
import pandas as pd
import numpy as np

marks = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [78, 85, 92, 67, 88]
})

print("Original Marks DataFrame:\n", marks)

# Calculate statistics for the 'Marks' column
mean_marks = marks['Marks'].mean()
median_marks = marks['Marks'].median()
max_marks = marks['Marks'].max()
min_marks = marks['Marks'].min()
std_dev_marks = marks['Marks'].std()
sum_marks = marks['Marks'].sum()

print(f"\nMean Marks: {mean_marks:.2f}")
print(f"Median Marks: {median_marks:.2f}")
print(f"Maximum Marks: {max_marks}")
print(f"Minimum Marks: {min_marks}")
print(f"Standard Deviation of Marks: {std_dev_marks:.2f}")
print(f"Sum of Marks: {sum_marks}")
```

### 17. Analyze student data by department: Average marks, maximum marks, number of students, and grouped statistics (Mean, Minimum, Maximum, Count).

```python
import pandas as pd

data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE"
    ],

    "Marks": [
        85, 78, 92,
        74, 88, 80
    ]
}

df = pd.DataFrame(data)
print("Original DataFrame:\n", df)

# Average marks by department
avg_marks_by_dept = df.groupby('Department')['Marks'].mean()
print("\nAverage marks by Department:\n", avg_marks_by_dept)

# Maximum marks by department
max_marks_by_dept = df.groupby('Department')['Marks'].max()
print("\nMaximum marks by Department:\n", max_marks_by_dept)

# Number of students in each department
students_per_dept = df['Department'].value_counts()
print("\nNumber of students in each Department:\n", students_per_dept)

# For each department, calculate: Mean, Minimum, Maximum, Count
dept_stats = df.groupby('Department')['Marks'].agg(['mean', 'min', 'max', 'count'])
print("\nDepartment-wise Statistics (Mean, Min, Max, Count):\n", dept_stats)
```

### 18. Combine two DataFrames using StudentID.

```python
import pandas as pd

students = pd.DataFrame({
    "StudentID": [101, 102, 103, 104],
    "Name": ["Asha", "Ravi", "Meena", "Kiran"]
})

marks = pd.DataFrame({
    "StudentID": [101, 102, 103, 104],
    "Marks": [85, 78, 92, 74]
})

print("Students DataFrame:\n", students)
print("\nMarks DataFrame:\n", marks)

# Combine both DataFrames using StudentID
combined_df = pd.merge(students, marks, on='StudentID')
print("\nCombined DataFrame:\n", combined_df)
```

### 20. Perform comprehensive analysis on an employee dataset, including statistics, filtering, sorting, and adding new categorical columns.

```python
import pandas as pd
import numpy as np

employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})

print("Original Employees DataFrame:\n", employees)

# Determine Average salary.
avg_salary = employees['Salary'].mean()
print(f"\nAverage Salary: ₹{avg_salary:,.2f}")

# Maximum and minimum salary.
max_salary = employees['Salary'].max()
min_salary = employees['Salary'].min()
print(f"Maximum Salary: ₹{max_salary:,.2f}")
print(f"Minimum Salary: ₹{min_salary:,.2f}")

# Employee receiving the highest salary.
highest_paid_employee = employees.loc[employees['Salary'].idxmax()]
print("\nEmployee receiving the highest salary:\n", highest_paid_employee)

# Employees earning above ₹60,000.
employees_above_60k = employees[employees['Salary'] > 60000]
print("\nEmployees earning above ₹60,000:\n", employees_above_60k)

# Average salary for each department.
avg_salary_by_dept = employees.groupby('Department')['Salary'].mean()
print("\nAverage Salary for each Department:\n", avg_salary_by_dept)

# Average experience by department.
avg_experience_by_dept = employees.groupby('Department')['Experience'].mean()
print("\nAverage Experience by Department:\n", avg_experience_by_dept)

# Number of employees in each department.
employees_per_dept = employees['Department'].value_counts()
print("\nNumber of Employees in each Department:\n", employees_per_dept)

# Sort employees by salary.
employees_sorted_by_salary = employees.sort_values(by='Salary', ascending=False)
print("\nEmployees sorted by Salary (Descending):\n", employees_sorted_by_salary)

# Find IT employees with more than 3 years of experience.
it_employees_experienced = employees[
    (employees['Department'] == 'IT') & (employees['Experience'] > 3)
]
print("\nIT Employees with more than 3 years of experience:\n", it_employees_experienced)

# Add a SalaryLevel column:
# ≥ ₹75,000 → High
# ₹50,000–₹74,999 → Medium
# < ₹50,000 → Low
conditions = [
    (employees['Salary'] >= 75000),
    (employees['Salary'] >= 50000) & (employees['Salary'] < 75000)
]
choices = ['High', 'Medium']
employees['SalaryLevel'] = np.select(conditions, choices, default='Low')
print("\nDataFrame with 'SalaryLevel' column:\n", employees)
```
