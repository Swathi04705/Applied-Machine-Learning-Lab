DIABETES DATASET

```python
data= pd.read_csv("drive/My Drive/Colab Notebooks/datasets/diabetes.csv")
dataset= data # loading data
data
```

```python

X = dataset.iloc[:, :8].values # Independent variables (all columns except 'Outcome')
y = dataset.iloc[:, 8].values    # Dependent variable ('Outcome' column)
```

```python
# Split the Diabetes dataset into training and testing sets
from sklearn.model_selection import train_test_split
X_train_diabetes, X_test_diabetes, y_train_diabetes, y_test_diabetes = train_test_split(X, y, test_size=0.25, random_state=42) # Using a different random_state

print("Shape of X_train_diabetes:", X_train_diabetes.shape)
print("Shape of X_test_diabetes:", X_test_diabetes.shape)
print("Shape of y_train_diabetes:", y_train_diabetes.shape)
print("Shape of y_test_diabetes:", y_test_diabetes.shape)
```

```python
# Apply Feature Scaling to the Diabetes dataset
from sklearn.preprocessing import StandardScaler

sc_diabetes = StandardScaler()
X_train_diabetes = sc_diabetes.fit_transform(X_train_diabetes)
X_test_diabetes = sc_diabetes.transform(X_test_diabetes)

print("First 5 rows of scaled X_train_diabetes:")
display(pd.DataFrame(X_train_diabetes).head())
print("First 5 rows of scaled X_test_diabetes:")
display(pd.DataFrame(X_test_diabetes).head())
```

```python
# Train a Decision Tree Classifier on the scaled Diabetes training data
from sklearn.tree import DecisionTreeClassifier

classifier_diabetes = DecisionTreeClassifier(criterion='gini', random_state=0, max_depth=5) # Limiting max_depth for better generalization
classifier_diabetes.fit(X_train_diabetes, y_train_diabetes)

print("Decision Tree Classifier trained for Diabetes dataset:")
display(classifier_diabetes)
```

```python
# Make predictions on the scaled Diabetes test set
Y_pred_diabetes = classifier_diabetes.predict(X_test_diabetes)

print("First 10 actual values (y_test_diabetes):", y_test_diabetes[:10])
print("First 10 predicted values (Y_pred_diabetes):", Y_pred_diabetes[:10])
```

```python
# Create and display the Confusion Matrix for the Diabetes dataset
from sklearn.metrics import confusion_matrix
cm_diabetes = confusion_matrix(y_test_diabetes, Y_pred_diabetes)

print("Confusion Matrix for Diabetes dataset:")
display(cm_diabetes)
```

```python
# Calculate and display the Accuracy Score for the Diabetes dataset
from sklearn import metrics
accuracy_diabetes = metrics.accuracy_score(y_test_diabetes, Y_pred_diabetes)

print(f"Accuracy for Diabetes dataset: {accuracy_diabetes * 100:.2f}%")
```

```python
# Display the Classification Report for the Diabetes dataset
from sklearn.metrics import classification_report

print("Classification Report for Diabetes dataset:")
print(classification_report(y_test_diabetes, Y_pred_diabetes))
```

```python
# Visualize the trained Decision Tree
from sklearn import tree
import matplotlib.pyplot as plt

plt.figure(figsize=(20,10))
tree.plot_tree(classifier_diabetes, filled=True, feature_names=dataset.drop('Outcome', axis=1).columns.tolist(), class_names=['No Diabetes', 'Diabetes'])
plt.title("Decision Tree for Diabetes Prediction")
plt.show()
```
